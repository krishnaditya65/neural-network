```python
print("Armstrong Numbers Between 1 and 1000:")
first = 100
last = 1000
while first<=last:
  res = 0
  temp = first
  noOfDigit = 0
  while temp>0:
    temp = int(temp/10)
    noOfDigit = noOfDigit + 1
  num = first
  while num>0:
    rem = num%10
    pow = 1
    i = 0
    while i<noOfDigit:
      pow = pow*rem
      i = i+1
    res = res+pow
    num = int(num/10)
  if res == first:
    print(res)
  first = first+1
```

```python
#PASS A LIST IN A FUNCTION AND FIND THE SUM, MAX,MIN VALUE OF THE NUMERIC ELEMENTS IN IT
def listfunc(lst):
    # Calculate sum of elements
    sum = 0
    for i in lst:
        sum += i

    # Find max value
    max1 = lst[0]
    for i in lst:
        if i > max1:
            max1 = i

    # Find min value
    min1 = lst[0]
    for i in lst:
        if i < min1:
            min1 = i

    # Return results as a tuple
    return (sum, max1, min1)

def main():
    list1=[5,10,15,20,25]
    result=listfunc(list1)
    print(result)
main()
    
```

```python
#PASS A LIST IN A FUNCTION AND RETURN THE SORTED LIST
def sort_list(my_list):
    n = len(my_list)
    for i in range(n):
        for j in range(i+1, n):
            if my_list[i] > my_list[j]:
                my_list[i], my_list[j] = my_list[j], my_list[i]
    return my_list

my_list = [5, 2, 8, 1, 9]
sorted_list = sort_list(my_list)
print(sorted_list)
```

```python
#PASS A TUPLE AND FIND THE MAXIMUM AND MINIMUM VALUE IN IT
def get_max_min(my_tuple):
    
    max_val = my_tuple[0]
    min_val = my_tuple[0]
    
    for val in my_tuple:
        if val > max_val:
            max_val = val
        elif val < min_val:
            min_val = val
    
    return (max_val, min_val)

my_tuple = (5, 2, 8, 1, 9)
max_val, min_val = get_max_min(my_tuple)
print("Maximum value:", max_val)
print("Minimum value:", min_val)
```

```python
#pass two numeric list in the function and find the sum and average of all elements in the list in python
def get_sum_avg(list1, list2):
    # Calculate the sum of all elements in the two lists
    total_sum = sum(list1) + sum(list2)
    
    # Calculate the average of all elements in the two lists
    total_count = len(list1) + len(list2)
    total_avg = total_sum / total_count
    
    # Return a tuple containing the sum and average values
    return (total_sum, total_avg)

list1 = [5, 2, 8, 1, 9]
list2 = [3, 7, 6, 4, 5]
total_sum, total_avg = get_sum_avg(list1, list2)
print("Total sum:", total_sum)
print("Total average:", total_avg)
```

```python
#pass a dictionary and return its keys and values in separate lists
def separate_dict(d):
    keys = []
    values = []
    for key in d:
        keys.append(key)
        values.append(d[key])
    return keys, values

my_dict = {'apple': 1, 'banana': 2, 'orange': 3}
keys, values = separate_dict(my_dict)
print(keys)  
print(values)
```

```python
#WRITE A FUNCTION TO GENERATE FIBONACCI NUMBERS BELOW 1000
def generate_fibonacci():
    fib = [1, 1]
    while fib[-1] < 1000:
        next_fib = fib[-1] + fib[-2]
        fib.append(next_fib)
    return fib[:-1]
fib = generate_fibonacci()
print(fib)
```

```python
#WRITE A FUNCTION TO GENERATE PRIME NUMBERS BELOW 100
def generate_primes():
    primes = []
    for num in range(2, 100):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes

primes = generate_primes()
print(primes)
```

```python
#RETURN THE ROOT OF A QUADRATIC EQUATION BY SUPPLYING THE VALUE OF PARAMETERS A,B,C.
def find_roots(a, b, c):
    discriminant = b**2 - 4*a*c
    if discriminant < 0:
        return "No real roots"
    elif discriminant == 0:
        return -b / (2*a)
    else:
        root1 = (-b + discriminant**0.5) / (2*a)
        root2 = (-b - discriminant**0.5) / (2*a)
        return root1, root2
    
print(find_roots(1, 5, 6))
```

```python
import math
x=int(input("enter x:"))
n=int(input("enter n:"))
def fact(x, n):
    result = x  # initialize result with x
    factorial = 1  # initialize factorial with 1
    for i in range(1, n+1):
        factorial *= i  # compute factorial(i)
        result += x / factorial  # add x/i! to the result
        x /= (x+i)  # update x for the next iteration
    return result
f=fact(x,n)
print(f)
```

```python
l=[12,33,44,55,66,77]
f=open("file1.bin",'wb')
for item in l:
    s=str(item)+"\n"
    bt=s.encode()
    f.write(bt)
f.close()
f=open("file1.bin",'rb')
for line in f:
    print(line)
    s=line.decode()
    print(s)
f.close()
```

```python
n=int(input("enter no of records"))
f=open("filerec.bin",'wb')
for i in range(n):
    roll=input("enter roll")
    name=input("enter name")
    age=input("enter age")
    mark=input("enter mark")
    s=roll+"\t"+name+"\t"+age+"\t"+mark+"\n"
    bt=s.encode()
    f.write(bt)
f.close()
f=open("filerec.bin",'rb')
for line in f:
    print(line)
    s=line.decode()
    print(s)
f.close()
```
